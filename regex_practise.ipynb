```python
import re
```

```python
with open('regex_corpus.txt', "r", encoding="utf-8") as fp:
    text = fp.read()
text
```

#### Q1: Match log lines that start with ERROR or WARN.

```python
pat = '(ERROR|WARN)'
re.findall(pat,text)
```

#### Q2: Match lines that start with TXN- and end with SUCCESS.

```python
pattern2 = re.compile(r'^TXN-.*SUCCESS', re.MULTILINE)
print("Transactions starting with TXN- and ending with SUCCESS:")
print(pattern2.findall(text), "\n")
```

#### Q3: Match all valid 10-digit phone numbers (ignore country code).

```python
pat3 = '[0-9]{10}'
re.findall(pat3,text)
```

#### Q4: Match any email where the username has exactly one dot (e.g. john.doe@...).

```python
pat4 = '\w+\.\w+@\w+.\w+'
re.findall(pat4,text)
```

#### Q5: Match the word cat but NOT catalog or scattered.

```python
re.findall('cat',text)
```

#### Q6: Find all valid emails in the text.

```python
pat6 = '\w+@\w+\.[a-z]{1,5}'
print(re.findall(pat6,text),"\n")
```

#### Q7: Find all prices in the text.

```python
pat7 = '₹\d+'
print(re.findall(pat7,text),"\n")
```
